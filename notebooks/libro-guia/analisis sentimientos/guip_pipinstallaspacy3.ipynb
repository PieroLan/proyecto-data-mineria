```python
import spacy
nlp = spacy.load("es_core_news_sm")

texto = "amando amarás amaba amor niños hablaremos"
doc = nlp(texto)
[(t.text, t.lemma_, t.pos_) for t in doc]
# Ej. -> [('amando','amar','VERB'), ('amarás','amar','VERB'), ('niños','niño','NOUN'), ...]
```

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import SnowballStemmer, WordNetLemmatizer

stemmer = SnowballStemmer("spanish")
lemmatizer = WordNetLemmatizer()

palabras = ["amando", "amarás", "amaba", "amor"]

print("Stemming:", [stemmer.stem(p) for p in palabras])
print("Lematización:", [lemmatizer.lemmatize(p, pos="v") for p in palabras])
```
