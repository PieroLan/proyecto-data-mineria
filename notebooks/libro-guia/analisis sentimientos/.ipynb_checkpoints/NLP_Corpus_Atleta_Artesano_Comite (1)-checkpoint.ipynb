# 🧠 Vectorización y Similitud con un Corpus Mini (Atleta–Artesano–Comité)
Corpus inspirado en el video (YouTube): *Doc 1 - atleta*, *Doc 2 - artesano*, *Doc 3 - comité*.
https://www.youtube.com/watch?v=4QsG4oNcGpM

Objetivos de la clase:
- Construir **BoW** y **TF-IDF** para un corpus pequeño
- Explorar el **vocabulario** y las **matrices** documento×término
- Medir **similitud coseno** entre documentos
- Visualizar resultados (mapa de calor de similitud)


## 1) Definir corpus

```python
docs = {
    "Doc 1": "atleta olímpica gana medalla olímpica",
    "Doc 2": "artesano diseña medalla olímpica",
    "Doc 3": "comité contrata artesano"
}

nombres = list(docs.keys())
corpus = list(docs.values())

for i, (k, v) in enumerate(docs.items(), 1):
    print(f"{i}. {k}: {v}")
```

## 2) Vectorización BoW (CountVectorizer)

```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

bow = CountVectorizer()
X_bow = bow.fit_transform(corpus)
vocab = bow.get_feature_names_out()

df_bow = pd.DataFrame(X_bow.toarray(), index=nombres, columns=vocab)
print("Vocabulario (BoW):", vocab.tolist())
df_bow
```

## 3) Vectorización TF-IDF (TfidfVectorizer)

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import numpy as np
tfidf = TfidfVectorizer()
X_tfidf = tfidf.fit_transform(corpus)
vocab_tfidf = tfidf.get_feature_names_out()

df_tfidf = pd.DataFrame(np.round(X_tfidf.toarray(), 3), index=nombres, columns=vocab_tfidf)
print("Vocabulario (TF-IDF):", vocab_tfidf.tolist())
df_tfidf
```

## 4) Similitud coseno entre documentos

```python
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Usamos TF-IDF para medir similitud (mejor que BoW en general)
S = cosine_similarity(X_tfidf)
S = np.round(S, 3)

import pandas as pd
df_sim = pd.DataFrame(S, index=nombres, columns=nombres)
df_sim
```

### 4.1) Mapa de calor de similitud (matplotlib)

```python
import matplotlib.pyplot as plt
import numpy as np

fig, ax = plt.subplots()
im = ax.imshow(S, vmin=0, vmax=1)  # colores por defecto
ax.set_xticks(np.arange(len(nombres)), labels=nombres, rotation=45, ha='right')
ax.set_yticks(np.arange(len(nombres)), labels=nombres)
ax.set_title("Similitud coseno entre documentos (TF-IDF)")
for i in range(S.shape[0]):
    for j in range(S.shape[1]):
        ax.text(j, i, S[i, j], ha='center', va='center')
plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
plt.tight_layout()
plt.show()
```

## 5) ¿Qué estamos viendo?
- **TF-IDF** resalta términos **informativos** como “atleta”, “artesano”, “comité” y “medalla”, “olímpica”.
- Los tres documentos comparten **“medalla olímpica”** → esto **aumenta la similitud** entre ellos.
- Aun así, cada documento contiene palabras que lo **diferencian** (atleta vs artesano vs comité).

## 6) (Extra) Bigramas para capturar frases

```python
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd, numpy as np
from sklearn.metrics.pairwise import cosine_similarity
# TF-IDF con unigramas + bigramas
tfidf_12 = TfidfVectorizer(ngram_range=(1,2))
X_12 = tfidf_12.fit_transform(corpus)
vocab_12 = tfidf_12.get_feature_names_out()

df_12 = pd.DataFrame(np.round(X_12.toarray(), 3), index=nombres, columns=vocab_12)
print("Incluye el bigrama 'medalla olímpica' si aparece:", any("medalla olímpica" == t for t in vocab_12))
display(df_12.loc[:, [t for t in vocab_12 if "medalla olímpica" in t]].head())

S12 = np.round(cosine_similarity(X_12), 3)
pd.DataFrame(S12, index=nombres, columns=nombres)
```
